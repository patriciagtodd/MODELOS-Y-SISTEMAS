# Ejercicio 1

Sea $\dot{x}(t) = Ax(t) + Bu(t)$ el sistema lineal invariante de orden 2, con condiciones iniciales $x_1(0) = x_2(0) = 1$, cuya matriz de transición está dada por

$$\Phi(t) = \begin{pmatrix} 2e^{-t} - e^{-2t} & e^{-t} - e^{-2t} \\ -2e^{-t} + 2e^{-2t} & -e^{-t} + 2e^{-2t} \end{pmatrix}.$$

### (a)

Mostrar que $A$ tiene autovalores $\lambda_1 = -1$, $\lambda_2 = -2$ y autovectores $(-1, 1)$, $(1, -2)$, respectivamente.

Hallar $\A$
Una propiedad fundamental es:
$$
\
A=\dot{\Phi}(0).
\
$$
Derivamos cada entrada de $\Phi(t)\$:
\
$$
\dot{\Phi}(t)=
\begin{pmatrix}
-2e^{-t}+2e^{-2t} & -e^{-t}+2e^{-2t}\\
2e^{-t}-4e^{-2t} & e^{-t}-4e^{-2t}
\end{pmatrix}.
\
$$
Ahora evaluamos en \(t=0\). Como \(e^0=1\),
$$
\
A=\dot{\Phi}(0)=
\begin{pmatrix}
0 & 1\\
-2 & -3
\end{pmatrix}.
\
$$
Chequeo rápido de los autovalores:
$$
\
\det(\lambda I-A)
=
\begin{vmatrix}
\lambda & -1\\
2 & \lambda+3
\end{vmatrix}
=
\lambda^2+3\lambda+2
=
(\lambda+1)(\lambda+2).
\
$$
Por lo tanto, $\lambda=-1 \ $  y  $ \lambda=-2 \ $, como pedía el inciso.

Graficar las variables de estado en función del tiempo para el sistema autónomo.
