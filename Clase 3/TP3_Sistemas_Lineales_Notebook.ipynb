# Ejercicio 1: explicación paso a paso

Se considera el sistema lineal invariante

$$
\dot{x}(t)=Ax(t)+Bu(t),
\qquad
x(0)=
\begin{pmatrix}
1\\
1
\end{pmatrix},
$$

con matriz de transición de estados

$$
\Phi(t)=
\begin{pmatrix}
2e^{-t}-e^{-2t} & e^{-t}-e^{-2t}\\
-2e^{-t}+2e^{-2t} & -e^{-t}+2e^{-2t}
\end{pmatrix}.
$$

La idea central es que, para el sistema autónomo (sin entrada, es decir, $u(t)=0$),

$$
x(t)=\Phi(t)x(0),
\qquad
\Phi(t)=e^{At}.
$$

La matriz $\Phi(t)$ toma el estado inicial $x(0)$ y lo transforma en el estado $x(t)$.

---

## 1. Hallar la matriz $A$

Una propiedad importante de la matriz de transición es:

$$
A=\dot{\Phi}(0).
$$

Derivamos cada entrada de $\Phi(t)$: OSEA LA JACOBIANA 

$$
\dot{\Phi}(t)=
\begin{pmatrix}
-2e^{-t}+2e^{-2t} & -e^{-t}+2e^{-2t}\\
2e^{-t}-4e^{-2t} & e^{-t}-4e^{-2t}
\end{pmatrix}.
$$

Al evaluar en $t=0$, se usa que $e^0=1$:

$$
A=\dot{\Phi}(0)=
\begin{pmatrix}
0 & 1\\
-2 & -3
\end{pmatrix}.
$$

---

## 2. Autovalores

Los autovalores se obtienen resolviendo:

$$
\det(\lambda I-A)=0.
$$

Primero:

$$
\lambda I-A=
\begin{pmatrix}
\lambda & -1\\
2 & \lambda+3
\end{pmatrix}.
$$

Entonces:

$$
\det(\lambda I-A)
=\lambda(\lambda+3)-(-1)(2)
=\lambda^2+3\lambda+2.
$$

Factorizamos:

$$
\lambda^2+3\lambda+2=(\lambda+1)(\lambda+2).
$$

Por lo tanto:

$$
\boxed{\lambda_1=-1,\qquad \lambda_2=-2.}
$$

---

## 3. Autovectores

Para cada autovalor buscamos un vector no nulo $v$ que cumpla:

$$
Av=\lambda v.
$$

Equivalentemente:

$$
(A-\lambda I)v=0.
$$

### Para $\lambda_1=-1$

$$
(A+I)v=0.
$$

$$
A+I=
\begin{pmatrix}
1 & 1\\
-2 & -2
\end{pmatrix}.
$$

Si $v=(x,y)^T$, la primera fila da:

$$
x+y=0
\quad\Rightarrow\quad
y=-x.
$$

Podemos elegir $x=-1$; entonces $y=1$. Por lo tanto:

$$
\boxed{v_1=
\begin{pmatrix}
-1\\
1
\end{pmatrix}.}
$$

### Para $\lambda_2=-2$

$$
(A+2I)v=0.
$$

$$
A+2I=
\begin{pmatrix}
2 & 1\\
-2 & -1
\end{pmatrix}.
$$

La primera fila da:

$$
2x+y=0
\quad\Rightarrow\quad
y=-2x.
$$

Elegimos $x=1$, por lo que $y=-2$. Así:

$$
\boxed{v_2=
\begin{pmatrix}
1\\
-2
\end{pmatrix}.}
$$

---

## 4. ¿De dónde salen los modos asociados?

No se adivinan: salen directamente de cada par **autovalor-autovector**.

Si $Av=\lambda v$, probamos una solución que se mantenga siempre en la dirección de $v$:

$$
x(t)=g(t)v.
$$

Al reemplazar en la ecuación $\dot{x}=Ax$:

$$
g'(t)v=A(g(t)v)=g(t)Av=g(t)\lambda v.
$$

Como $v\neq0$:

$$
g'(t)=\lambda g(t).
$$

La solución de esa ecuación escalar es:

$$
g(t)=ce^{\lambda t}.
$$

Por eso, para cada autovalor y autovector, aparece un modo de la forma:

$$
e^{\lambda t}v.
$$

En este ejercicio, los modos son:

$$
\boxed{
e^{-t}
\begin{pmatrix}-1\\1\end{pmatrix}}
\qquad\text{y}\qquad
\boxed{
e^{-2t}
\begin{pmatrix}1\\-2\end{pmatrix}.}
$$

El autovector indica la dirección del movimiento; el autovalor indica cómo cambia la magnitud con el tiempo. Como ambos autovalores son negativos, ambos modos se van apagando cuando $t$ aumenta.

La solución general es una combinación lineal de esos dos modos:

$$
\boxed{
x(t)=c_1e^{-t}
\begin{pmatrix}-1\\1\end{pmatrix}
+c_2e^{-2t}
\begin{pmatrix}1\\-2\end{pmatrix}.}
$$

---

## 5. Usar la condición inicial

Para hallar $c_1$ y $c_2$, imponemos $x(0)=(1,1)^T$:

$$
\begin{pmatrix}1\\1\end{pmatrix}
=c_1
\begin{pmatrix}-1\\1\end{pmatrix}
+c_2
\begin{pmatrix}1\\-2\end{pmatrix}.
$$

Esto da el sistema:

$$
\begin{cases}
-c_1+c_2=1\\
c_1-2c_2=1.
\end{cases}
$$

Su solución es:

$$
c_1=-3,
\qquad
c_2=-2.
$$

Entonces:

$$
x(t)=-3e^{-t}
\begin{pmatrix}-1\\1\end{pmatrix}
-2e^{-2t}
\begin{pmatrix}1\\-2\end{pmatrix}.
$$

Al hacer las cuentas:

$$
\boxed{
x(t)=
\begin{pmatrix}
3e^{-t}-2e^{-2t}\\
-3e^{-t}+4e^{-2t}
\end{pmatrix}.}
$$

Por lo tanto:

$$
x_1(t)=3e^{-t}-2e^{-2t},
\qquad
x_2(t)=-3e^{-t}+4e^{-2t}.
$$

También se puede llegar exactamente al mismo resultado multiplicando directamente:

$$
x(t)=\Phi(t)x(0).
$$

---

## 6. Interpretación y gráfico del caso autónomo

Como $\lambda_1=-1$ y $\lambda_2=-2$ son negativos:

$$
e^{-t}\to0,
\qquad
e^{-2t}\to0
\qquad (t\to\infty).
$$

Por eso, sin entrada, ambas variables de estado terminan tendiendo a cero.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 500)

x1 = 3*np.exp(-t) - 2*np.exp(-2*t)
x2 = -3*np.exp(-t) + 4*np.exp(-2*t)

plt.plot(t, x1, label='x1(t)')
plt.plot(t, x2, label='x2(t)')
plt.axhline(0, color='black', linewidth=0.7)
plt.xlabel('t')
plt.ylabel('Estados')
plt.grid()
plt.legend()
plt.show()
```

## 7. Caso no homogéneo: $u(t)=1$

Ahora se tiene:

$$
B=
\begin{pmatrix}
0\\
1
\end{pmatrix},
\qquad u(t)=1.
$$

La solución de un sistema lineal con entrada es:

$$
x(t)=\Phi(t)x(0)+
\int_0^t\Phi(t-\tau)B u(\tau)\,d\tau.
$$

La primera parte es la solución homogénea, que ya conocemos. Para la integral, como $B=(0,1)^T$, $\Phi(s)B$ es la segunda columna de $\Phi(s)$:

$$
\Phi(s)B=
\begin{pmatrix}
e^{-s}-e^{-2s}\\
-e^{-s}+2e^{-2s}
\end{pmatrix}.
$$

Al integrar entre $0$ y $t$:

$$
\int_0^t\Phi(s)B\,ds=
\begin{pmatrix}
\frac12-e^{-t}+\frac12e^{-2t}\\
e^{-t}-e^{-2t}
\end{pmatrix}.
$$

Sumando la parte homogénea y la parte producida por la entrada:

$$
\boxed{
x(t)=
\begin{pmatrix}
\frac12+2e^{-t}-\frac32e^{-2t}\\
-2e^{-t}+3e^{-2t}
\end{pmatrix}.}
$$

En este caso:

$$
\lim_{t\to\infty}x(t)=
\begin{pmatrix}
\frac12\\
0
\end{pmatrix}.
$$

La entrada constante evita que el estado termine en el origen.
