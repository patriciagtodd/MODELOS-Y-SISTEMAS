### (a)

$$
\dot{x} = \sqrt{x},
\qquad x(0) = 2,
\qquad t_0 = 2
$$

```python
import numpy as np
import pandas as pd

def f(x):
    return np.sqrt(x)

def euler_method(f, x0, t_inicial, t_final, h):
    num_steps = int(round((t_final - t_inicial) / h))
    tt = np.linspace(t_inicial, t_final, num_steps + 1)
    xx = np.zeros(num_steps + 1)
    xx[0] = x0

    for i in range(num_steps):
        xx[i+1] = xx[i] + h * f(xx[i])

    return tt, xx

x0 = 2
t = np.array([0,5])
t0 = 2

hh = [0.5, 0.1, 0.05]

resultados = []

for h in hh:
    tt, xx = euler_method(f, x0, t[0], t[-1], h)

    idx_target = np.argmin(np.abs(tt - t0))
    x_at_target = xx[idx_target]

    resultados.append({'h': h, 'x(t_0)': x_at_target, 't_at_target': tt[idx_target]})

results_df = pd.DataFrame(resultados)

print(f"Solución para el ejercicio 4(a) usando el método de Euler:")
print(f"Ecuación: dx/dt = sqrt(x), x(0) = {x0}, t_0 = {t0}")
print("\nResultados de la simulación para diferentes pasos h:")
print(results_df.to_markdown(index=False))
```

### (b)

$$
\begin{cases}
\dot{x} = y \\
\dot{y} = x-y
\end{cases}
\qquad
x(0)=1,\quad y(0)=2,
\qquad t_0=1
$$

```python
import numpy as np
import pandas as pd

# Definimos la función del sistema de EDOs para el ejercicio 4(b)
# El estado es un vector [x, y], y f(estado, t) retorna [dx/dt, dy/dt]
def system_ode_b(state_vector, t):
    x, y = state_vector
    dxdt = y
    dydt = x - y
    return np.array([dxdt, dydt])

# Implementación genérica del método de Euler para sistemas de EDOs
# f: función que describe el sistema de EDOs (vectorial)
# x0: vector de condiciones iniciales [x1(0), x2(0), ...]
# t_inicial, t_final: rango de tiempo de simulación
# h: tamaño del paso
def euler_method_system(f, x0, t_inicial, t_final, h):
    # Calculamos el número de pasos necesario
    num_steps = int(round((t_final - t_inicial) / h))
    # Generamos los valores de tiempo
    tt = np.linspace(t_inicial, t_final, num_steps + 1)
    # Inicializamos el array para almacenar los estados (filas: tiempo, columnas: variables de estado)
    xx = np.zeros((num_steps + 1, len(x0)))
    xx[0] = x0

    # Iteramos para aplicar el método de Euler
    for i in range(num_steps):
        # x_{i+1} = x_i + h * f(x_i, t_i)
        xx[i+1] = xx[i] + h * f(xx[i], tt[i])

    return tt, xx

# Condiciones iniciales y parámetros para el ejercicio 4(b)
valores_iniciales = np.array([1, 2]) # [x(0), y(0)]
tt = np.array([0, 5]) # Rango de tiempo para la simulació
t0 = 1 # El tiempo t0 en el que queremos encontrar x e y

# Distintos tamaños de paso a simular
hh = [0.5, 0.1, 0.05]

results = []

for h in hh:
    # ejecuto metodo euler
    tt, xx = euler_method_system(system_ode_b, valores_iniciales, tt[0], tt[-1], h)

    # busco el indice mas cercano a t0
    idx_t0 = np.argmin(np.abs(tt - t0))

    # Obtenemos los valores de x e y en t_target
    x_t0 = xx[idx_t0, 0]
    y_t0 = xx[idx_t0, 1]

    results.append({'h': h, 'x(t_0)': x_t0, 'y(t_0)': y_t0, 't_at_target': tt[idx_t0]})

# Creamos un DataFrame para mostrar los resultados de forma tabular
results_df = pd.DataFrame(results)

print(f"Solución para el ejercicio 4(b) usando el método de Euler para sistemas:")
print(f"Ecuación: dx/dt = y, dy/dt = x-y, x(0) = {valores_iniciales[0]}, y(0) = {valores_iniciales[1]}, t_0 = {t0}")
print("\nResultados de la simulación para diferentes pasos h:")
print(results_df.to_markdown(index=False))
```
